```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
customers = pd.read_csv("../../data/raw/customers_medium.csv")
menu_items = pd.read_csv("../../data/raw/menu_items.csv")
order_items = pd.read_csv("../../data/raw/order_items (2).csv")
orders = pd.read_csv("../../data/raw/orders_medium.csv")
restaurants = pd.read_csv("../../data/raw/restaurants.csv")
```

```python
customers_clean = customers.copy()
menu_items_clean = menu_items.copy()
order_items_clean = order_items.copy()
orders_clean = orders.copy()
restaurants_clean = restaurants.copy()
```

```python
datasets_clean = {
    "Customers": customers_clean,
    "Menu Items": menu_items_clean,
    "Order Items": order_items_clean,
    "Orders": orders_clean,
    "Restaurants": restaurants_clean
}

for name, df in datasets_clean.items():
    print("===")
    print(name)
    print("===")
   
    print(df.dtypes)
```

```python
customers_clean["signup_date"] = pd.to_datetime(
    customers_clean["signup_date"]
)

orders_clean["order_time"] = pd.to_datetime(
    orders_clean["order_time"]
)

orders_clean["delivery_time"] = pd.to_datetime(
    orders_clean["delivery_time"]
)
```

```python
print(customers_clean["signup_date"].dtype)
print(orders_clean["order_time"].dtype)
print(orders_clean["delivery_time"].dtype)
```

```python
datasets_clean = {
    "Customers": customers_clean,
    "Menu Items": menu_items_clean,
    "Order Items": order_items_clean,
    "Orders": orders_clean,
    "Restaurants": restaurants_clean
}

for name, df in datasets_clean.items():
    print(f"\n{'='*50}")
    print(name)
    print('='*50)
    print(df.dtypes)
```

```python
invalid_delivery = orders_clean[
    orders_clean["delivery_time"] < orders_clean["order_time"]
]

print("Invalid delivery records:", len(invalid_delivery))
```

```python
print("Negative menu prices:",
      (menu_items_clean["price"] < 0).sum())

print("Negative order item prices:",
      (order_items_clean["price"] < 0).sum())
```

```python
print("Invalid quantities:",
      (order_items_clean["quantity"] <= 0).sum())
```

```python
print("Invalid ratings:",
      ((restaurants_clean["rating"] < 0) |
       (restaurants_clean["rating"] > 5)).sum())
```

```python
print("Customer signup range:")
print(
    customers_clean["signup_date"].min(),
    "to",
    customers_clean["signup_date"].max()
)

print("\nOrder date range:")
print(
    orders_clean["order_time"].min(),
    "to",
    orders_clean["order_time"].max()
)
```
